#**Problema 1:**
Se hace un estudio sobre la efectividad de tres marcas de atomizador para matar moscas. Para ello, cada producto se aplica a un grupo de 100 moscas, y se cuenta el número de moscas muertas expresado en porcentajes. Se hicieron seis réplicas, pero en días diferentes; por ello, se sospecha que puede haber algún efecto importante debido a esta fuente de variación. Los datos obtenidos se muestran a continuación:

| Marca de atomizador | Réplica 1 | Réplica 2 | Réplica 3 | Réplica 4 | Réplica 5 | Réplica 6 |
|----------------------|-----------|-----------|-----------|-----------|-----------|-----------|
| 1                   | 72        | 65        | 67        | 75        | 62        | 73        |
| 2                   | 55        | 59        | 68        | 70        | 53        | 50        |
| 3                   | 64        | 74        | 61        | 58        | 51        | 69        |

**Preguntas:**<br>
a) Suponiendo un DBCA, formule las hipótesis adecuadas y el modelo estadístico.  
b) ¿Existe diferencia entre la efectividad promedio de los atomizadores?  
c) ¿Hay algún atomizador mejor? Argumente su respuesta.  
d) ¿Hay diferencias significativas en los resultados de diferentes días en que se realizó el experimento? Argumente su respuesta.  
e) Verifique los supuestos de normalidad y de igual varianza entre las marcas.


````python
# csv es un acrónimo de "comma separated values" o "valores separados por comas".
# Lo primero que hay que hacer es pasar estos datos a una tabla de valores crudos
# en formato csv. Recuerden el problema que fue usar espacios en el csv, por esta
# razón, evítenlos.

import pandas as pd
from io import StringIO

datos = '''
marca,replica,moscas
M1,R1,72
M1,R2,65
M1,R3,67
M1,R4,75
M1,R5,62
M1,R6,73
M2,R1,55
M2,R2,59
M2,R3,68
M2,R4,70
M2,R5,53
M2,R6,50
M3,R1,64
M3,R2,74
M3,R3,61
M3,R4,58
M3,R5,51
M3,R6,69
'''

df = pd.read_csv(StringIO(datos))
df.reset_index(drop=True, inplace=True)

df

# JOCELYN GUADALUPE CARREÓN RANGEL
````

````python
#ANOVA PARA CHECAR SI HAY DIFERENCIA DE MEDIAS


from statsmodels.formula.api import ols
import statsmodels.api as sm

# Nivel de significancia = 0.05
# H0: Todas las medias son iguales
# H1: Al menos una media es distinta

# Realizar el modelo ANOVA de dos vías
modelo_lineal = ols(f'moscas ~ C(marca)+C(replica)', data=df).fit()
tabla_anova = sm.stats.anova_lm(modelo_lineal, typ=2)
tabla_anova

#replica es el bloqueo, no nos interesa

# JOCELYN GUADALUPE CARREÓN RANGEL
````

**H0:** Todas las replicas/marcas son iguales

**H1:** Al menos una replica/marca es distinta


```
El valor p es mayor a 0.05 entoces, H0 es verdadera.
```
Por lo tanto, **todas las medias son iguales (marcas y replicas)**.

````python
#SHAPIRO PARA CHECAR LA NORMALIDAD

from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns

residuos = modelo_lineal.resid

# Prueba de Shapiro-Wilk
stat, valor_p_sh = shapiro(residuos)
print(f"valor-p (Shapiro) = {valor_p_sh}")

# Visualización: Q-Q plot
sm.qqplot(residuos, line='s')
plt.title("Q-Q plot de residuos")
plt.show()

# Histograma
sns.histplot(residuos, kde=True)
plt.title("Histograma de residuos")
plt.show()

# JOCELYN GUADALUPE CARREÓN RANGEL
````

El valor p es de 0.55 y es mayor al nivel de significancia, por lo tanto, **si hay normalidad**.

````python
#LEVENE PARA HOMOCEDASTICIDAD

#H0: Homocedasticidad
#H1: Heterocedasticidad

from scipy.stats import levene
# Supón que tienes un DataFrame df con columnas 'respuesta' y 'grupo'
grupo1 = df[df['marca'] == 'M1']['moscas']
grupo2 = df[df['marca'] == 'M2']['moscas']
grupo3 = df[df['marca'] == 'M3']['moscas']


stat, p = levene(grupo1, grupo2, grupo3)
print(f"p-valor de Levene: {p}")

# JOCELYN GUADALUPE CARREÓN RANGEL
````

El valor p es mayor a 0.05, por lo tato, **si hay homocedasticidad**.

````python
#TUKEY PARA COMPARAR QUE GRUPO ES MEJOR

from statsmodels.stats.multicomp import pairwise_tukeyhsd
import matplotlib.pyplot as plt

nivel_de_significancia = 0.05

# Prueba de Tukey
tukey = pairwise_tukeyhsd(endog=df['moscas'], groups=df['marca'], alpha=nivel_de_significancia)

# Mostrar los resultados
print(tukey)

# Gráfico de las diferencias entre grupos
tukey.plot_simultaneous(ylabel="Marcas", xlabel="Moscas")

plt.gca().spines['right'].set_visible(False) # derecha
plt.gca().spines['top'].set_visible(False)   # superior
plt.title("Múltiples comparaciones entre todos los pares (Tukey)")

# JOCELYN GUADALUPE CARREÓN RANGEL
````

El Tukey nos dice que dice que **las marcas no tienen tanta diferencia**.
Por lo tanto, todas las marcas de rociadores son recomendables.

**False= no hay diferencia**

````python
#ANOVA PARA CHECAR SI HAY DIFERENCIA DE MEDIAS


from statsmodels.formula.api import ols
import statsmodels.api as sm

# Nivel de significancia = 0.05
# H0: Todas las medias son iguales
# H1: Al menos una media es distinta

# Realizar el modelo ANOVA de dos vías
modelo_lineal = ols(f'temperatura ~ C(silos)+C(dia)', data=df).fit()
tabla_anova = sm.stats.anova_lm(modelo_lineal, typ=2)
tabla_anova

#replica es el bloqueo, no nos interesa

# JOCELYN GUADALUPE CARREÓN RANGEL
````

**H0:** Todas los silos/dias son iguales

**H1:** Al menos un silos/dias es distinta


```
El valor p es mayor a 0.05 entoces, H0 es verdadera.
```
Por lo tanto, **todas las medias son iguales (silos y dia)**.

````python
#SHAPIRO PARA CHECAR LA NORMALIDAD

from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns

residuos = modelo_lineal.resid

# Prueba de Shapiro-Wilk
stat, valor_p_sh = shapiro(residuos)
print(f"valor-p (Shapiro) = {valor_p_sh}")

# Visualización: Q-Q plot
sm.qqplot(residuos, line='s')
plt.title("Q-Q plot de residuos")
plt.show()

# Histograma
sns.histplot(residuos, kde=True)
plt.title("Histograma de residuos")
plt.show()

# JOCELYN GUADALUPE CARREÓN RANGEL
````

El valor p es de 0.1527 y es mayor al nivel de significancia, por lo tanto, **si hay normalidad**.

````python
#LEVENE PARA HOMOCEDASTICIDAD

#H0: Homocedasticidad
#H1: Heterocedasticidad

from scipy.stats import levene
# Supón que tienes un DataFrame df con columnas 'respuesta' y 'grupo'
grupo1 = df[df['silos'] == 'A']['temperatura']
grupo2 = df[df['silos'] == 'B']['temperatura']
grupo3 = df[df['silos'] == 'C']['temperatura']
grupo4 = df[df['silos'] == 'D']['temperatura']
grupo5 = df[df['silos'] == 'E']['temperatura']


stat, p = levene(grupo1, grupo2, grupo3, grupo4, grupo5)
print(f"p-valor de Levene: {p}")

# JOCELYN GUADALUPE CARREÓN RANGEL
````

El valor p es mayor a 0.05, por lo tato, **si hay homocedasticidad**.

Como el ANOVA nos dice que todas las medias son iguales, no es necesario hacer Tukey ya que nos va a decir lo mismo, que **los silos no tienen tanta diferencia**.
Por lo tanto, todas las marcas de rociadores son recomendables.

**False= no hay diferencia**

````python
#TUKEY PARA COMPARAR QUE GRUPO ES MEJOR

from statsmodels.stats.multicomp import pairwise_tukeyhsd
import matplotlib.pyplot as plt

nivel_de_significancia = 0.05

# Prueba de Tukey
tukey = pairwise_tukeyhsd(endog=df['temperatura'], groups=df['silos'], alpha=nivel_de_significancia)

# Mostrar los resultados
print(tukey)

# Gráfico de las diferencias entre grupos
tukey.plot_simultaneous(ylabel="Silos", xlabel="Temperatura")

plt.gca().spines['right'].set_visible(False) # derecha
plt.gca().spines['top'].set_visible(False)   # superior
plt.title("Múltiples comparaciones entre todos los pares (Tukey)")

# JOCELYN GUADALUPE CARREÓN RANGEL
````

Deciden correr un experimento en cuadro grecolatino controlando los factores de bloque **chofer**, marca de vehículo (**α, β, χ, δ**) y día de la semana. El experimento se repite en dos semanas diferentes, en las cuales no hay días festivos ni quincenas. Los costos observados en pesos se muestran en la siguiente tabla:

| Chofer/día | Lunes       | Martes      | Miércoles   | Jueves      |
|------------|-------------|-------------|-------------|-------------|

|            | 750         | 610         | 580         | 650         |
| Enrique    | 650(A, χ)   | 540(B, α)   | 580(C, χ)   | 850(D, β)   |
|            | 725         | 560         | 635         | 770         |
| Genaro     | 700(C, β)   | 650(D, δ)   | 635(A, α)   | 450(B, χ)   |
|            | 675         | 740         | 540         | 550         |
| Luis       | 475(B, δ)   | 560(A, β)   | 650(D, α)   | 670(C, α)   |
|            | 480         | 615         | 725         | 730         |

### Preguntas:
a) Haga el análisis de varianza de este experimento.

h) Verifique los supuestos del modelo.  


````python
# csv es un acrónimo de "comma separated values" o "valores separados por comas".
# Lo primero que hay que hacer es pasar estos datos a una tabla de valores crudos
# en formato csv. Recuerden el problema que fue usar espacios en el csv, por esta
# razón, evítenlos.

import pandas as pd
from io import StringIO

datos = '''
chofer,dia,ruta,marca,costo
Carlos,Lun,D,alpha,825
Carlos,Lun,D,alpha,750
Carlos,Mar,C,gamma,585
Carlos,Mar,C,gamma,610
Carlos,Mier,B,beta,550
Carlos,Mier,B,beta,580
Carlos,Jue,A,delta,580
Carlos,Jue,A,delta,650
Enrique,Lun,A,gamma,650
Enrique,Lun,A,gamma,725
Enrique,Mar,B,alpha,540
Enrique,Mar,B,alpha,560
Enrique,Mier,C,gamma,580
Enrique,Mier,C,gamma,635
Enrique,Jue,D,beta,850
Enrique,Jue,D,beta,770
Genaro,Lun,C,beta,700
Genaro,Lun,C,beta,675
Genaro,Mar,D,delta,650
Genaro,Mar,D,delta,740
Genaro,Mier,A,alpha,635
Genaro,Mier,A,alpha,540
Genaro,Jue,B,gamma,450
Genaro,Jue,B,gamma,550
Luis,Lun,B,delta,475
Luis,Lun,B,delta,480
Luis,Mar,A,beta,560
Luis,Mar,A,beta,615
Luis,Mier,D,alpha,650
Luis,Mier,D,alpha,725
Luis,Jue,C,alpha,670
Luis,Jue,C,alpha,730
 '''
df = pd.read_csv(StringIO(datos))
df.reset_index(drop=True, inplace=True)

df

# JOCELYN GUADALUPE CARREÓN RANGEL
````

````python
#ANOVA PARA CHECAR SI HAY DIFERENCIA DE MEDIAS


from statsmodels.formula.api import ols
import statsmodels.api as sm

# Nivel de significancia = 0.05
# H0: Todas las medias son iguales
# H1: Al menos una media es distinta

# Realizar el modelo ANOVA de dos vías
modelo_lineal = ols(f'costo ~ C(chofer)+C(dia)+C(ruta)+C(marca)', data=df).fit()
tabla_anova = sm.stats.anova_lm(modelo_lineal, typ=2)
tabla_anova

#replica es el bloqueo, no nos interesa

# JOCELYN GUADALUPE CARREÓN RANGEL
````

- chofer > 0.05

**H0: el chofer no influye en el costo.**

H1: al menos un chofer influye significativamente en el costo.

- día < 0.05

H0: el costo es el mismo sin importar el día de la semana.

**H1: el costo varía significativamente según el día.**

- ruta < 0.05

H0: el costo promedio es igual en las cuatro rutas.

**H1: al menos una ruta tiene un costo promedio diferente.**

- marca > 0.05

**H0: la marca del vehiculo no afecta el consumo ni el tiempo.**

H1: al menos una marca de vehículo genera costos diferentes.

````python
#SHAPIRO PARA CHECAR LA NORMALIDAD

from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns

residuos = modelo_lineal.resid

# Prueba de Shapiro-Wilk
stat, valor_p_sh = shapiro(residuos)
print(f"valor-p (Shapiro) = {valor_p_sh}")

# Visualización: Q-Q plot
sm.qqplot(residuos, line='s')
plt.title("Q-Q plot de residuos")
plt.show()

# Histograma
sns.histplot(residuos, kde=True)
plt.title("Histograma de residuos")
plt.show()

# JOCELYN GUADALUPE CARREÓN RANGEL
````

El valor p del test de Shapiro es mayor al nivel de significancia.

Por lo tanto, la hipotesis nula es verdadera, **si hay normalidad**.

````python
#LEVENE PARA HOMOCEDASTICIDAD

#H0: Homocedasticidad
#H1: Heterocedasticidad

from scipy.stats import levene

# Supón que tienes un DataFrame df con columnas 'respuesta' y 'grupo'
grupo1 = df[df['ruta'] == 'A']['costo']
grupo2 = df[df['ruta'] == 'B']['costo']
grupo3 = df[df['ruta'] == 'C']['costo']
grupo4 = df[df['ruta'] == 'D']['costo']

stat, p = levene(grupo1, grupo2, grupo3, grupo4)
print(f"p-valor de Levene: {p}")

# JOCELYN GUADALUPE CARREÓN RANGEL
````

El valor p es mayor al nivel de significancia.

Por lo tanto, **si hay homocedasticidad**.

````python
#       PASO 5 (4) TUKEY  PARA SABER CUAL ES MEJOR O MAS VARIABLE
#   referencia visual

from statsmodels.stats.multicomp import pairwise_tukeyhsd
import matplotlib.pyplot as plt

nivel_de_significancia = 0.05

# Prueba de Tukey
tukey = pairwise_tukeyhsd(endog=df['costo'], groups=df['ruta'], alpha=nivel_de_significancia)

# Mostrar los resultados
print(tukey)

# Gráfico de las diferencias entre grupos
tukey.plot_simultaneous(ylabel="ruta", xlabel="costos")

plt.gca().spines['right'].set_visible(False) # derecha
plt.gca().spines['top'].set_visible(False)   # superior
plt.title("Múltiples comparaciones entre todos los pares (Tukey)")

# JOCELYN GUADALUPE CARREÓN RANGEL
````
